```python
#----------------The Network--------------
"""
Base Model: F. Escudero
Structural Modifications: A. Schwarz (2018)
Version Control & Algorithmic Optimization: K. Yahya (2026)
Current Version: 2.1
Date: 30.03.2026
"""
# from ANNarchy import *
from ANNarchy import (
    Constant,
    Neuron,
    Synapse,
    Population,
    Projection,
    get_population,
    get_projection,
    Uniform,
    Monitor,
    compile,
    setup,
    reset,
    clear,
    projections,

)
#from parameters import params
import numpy as np
from time import time
from CompNeuroPy import DBSstimulator, compile_in_folder
import re

############################# clear network ##################################
clear()

##############################################################################
################################# init variables #############################
##############################################################################

baseline_dopa = Constant("baseline_dopa", params["baseline_dopa"])


def create_network(
    seed, dbs_state, shortcut, parameter, dbs_param_state, ann_compile_str="annarchy"
):
    ######################### Setup ANNarchy function ############################

    # load seed parameters
    setup(seed=seed)

    ##############################################################################
    ################################# neurons ####################################
    ##############################################################################

    ################################# input ######################################

    # PPN, IT
    InputNeuron = Neuron(
        parameters="""
            tau = 10.0 : population
            B = 0.0
        """,
        equations="""
            tau * dmp/dt = -mp + B
            r = pos(mp)
        """,
        name="Input Neuron",
        description="Rate-coded neuron only with a baseline to be set.",
    )

    ################################# standard ####################################
    # StrD1, StrD2, STN, GPi, GPe, Thal, PFC, StrThal, MTL
    LinearNeuron = Neuron(
        parameters="""
            tau = 10.0 : population
            phi = 0.0 : population
            B = 0.0
            input = 0.0
        """,
        equations="""
            tau * dmp/dt = -mp + sum(exc) - sum(inh) + B + input + phi * Uniform(-1.0,1.0)
            r = pos(mp)
        """,
        name="Linear Neuron",
        description="Regular rate-coded neuron with excitatory and inhibitory inputs plus baseline and noise.",
    )
    MemoryNeuron = Neuron(
    parameters="""
        tau = 50.0 : population   # افزایش زمان برای دیرتر خاموش شدن
        phi = 0.05 : population  # نویز کمتر برای پایداری
        B = 0.0
        input = 0.0
        adaptation = 0.01        # اختیاری: برای اینکه تا ابد روشن نمونه
    """,
    equations="""
        # اضافه کردن یک ترم بازگشتی یا خود-تحریکی در sum(exc)
        tau * dmp/dt = -mp + sum(exc) - sum(inh) + B + input + phi * Uniform(-1.0,1.0)

        # خروجی نورون
        r = pos(mp)
    """,
    name="Memory Neuron",
    )

    ################################# Dopamin ######################################
    # SNc
    DopamineNeuron = Neuron(
        parameters="""
            tau = 10.0 : population
            alpha = 0 : population
            B = 0.0
        """,
        equations="""
            aux = if (sum(exc)>0): pos(1.0-B-sum(inh)) else: -10 * sum(inh)
            tau * dmp/dt = -mp + alpha * aux + B
            r = pos(mp)
        """,
        name="Dopamine Neuron",
        description="Excitatory input increases activity above constant baseline. Inhibitory input can prevent the increase.",
    )

    ##############################################################################
    ################################# synapses ###################################
    ##############################################################################

    ########################## standard synapses (DA) #############################
    # PPNSNc, SNcStrD1, SNcStrD2, SNcSTN, SNcGPe, SNcGPi
    StandardSynapse = Synapse(
        psp="w * pre.r",
        name="Standard",
        description="Standard synapse, without plasticity which calculates the psp as a multiplication of weight and pre-synaptic rate.",
    )

    # ITThal
   # PostCovariance = Synapse(
       # parameters="""
         #   tau = 70.0 : projection
          #  tau_alpha = 1.0 : projection
         #   regularization_threshold = 0.5: projection
           # threshold_post = 0.1 : projection
          #  threshold_pre = 0.0 : projection
          #  alpha_factor = 2.0 : projection
            #12=1e-4 : projection
      #  """,
     #   psp="w * pre.r",
       # equations="""
           # tau_alpha * dalpha/dt  + alpha =  pos(post.mp - regularization_threshold) * alpha_factor
          #  trace = pos(pre.r - threshold_pre) * (post.r - mean(post.r) - threshold_post)
           # delta = (trace - alpha*pos(post.r - mean(post.r) - threshold_post) * pos(post.r - mean(post.r) - threshold_post)*w)
           # tau * dw/dt = deltas : min = 0, max=0.6
   # """,
       # name="Covariance learning rule",
       # description="Synaptic plasticity based on covariance, with an additional regularization term.",
 #   )

    PostCovariance = Synapse(
    parameters="""
        tau = 15000.0
        tau_alpha = 10.0
        regularization_threshold = 1.0
        threshold_post = 0.0
        threshold_pre = 0.0
    """,
    equations="""
        # محاسبه میانگین فعالیت برای تنظیم وزن
        tau_alpha * dalpha/dt + alpha = pos(post.r - regularization_threshold)

        # قانون کوواریانس
        trace = (pre.r - mean(pre.r) - threshold_pre) * pos(post.r - mean(post.r) - threshold_post)

        # محاسبه تغییرات نهایی
        delta = trace - alpha * w
        tau * dw/dt = delta : min=0
    """,
    name="Covariance_learning_rule",
    description="Synaptic plasticity based on covariance with regularization"
    )

    ############################## striatal synapse ##################################
    # ITStrD1, ITStrD2, ITSTN
    DAPostCovarianceNoThreshold = Synapse(
        parameters="""
            tau=70.0 : projection
            tau_alpha=1.0 : projection
            tau_trace=60.0 : projection
            regularization_threshold=1.0 : projection
            K_burst = 1.0 : projection
            K_dip = 0.4 : projection
            DA_type = -1 : projection
            threshold_pre=0.005 : projection #0.03
            threshold_post=0.05 : projection #0.15
        """,
        psp="w * pre.r",
        equations="""
            tau_alpha * dalpha/dt + alpha = pos(post.mp - regularization_threshold)
            dopa_sum = 2.0 * (post.sum(dopa) - baseline_dopa)
            #trace = pos(post.r - mean(post.r)- threshold_post) * pre.r
            tau_trace * dtrace/dt + trace = pos(post.r - mean(post.r) - threshold_post) * pre.r
            dopa_mod =  if (DA_type*dopa_sum>0):
                            DA_type*K_burst*dopa_sum
                        else:
                            if (trace>0.0) and (w >0.0):
                                DA_type*K_dip*dopa_sum
                            else:
                                0

            delta = (dopa_mod* trace - alpha*pos(post.r - mean(post.r) - threshold_post)*pos(post.r - mean(post.r) - threshold_post))

            tau * dw/dt = delta : min=0 , max=0.6
        """,
        name="Covariance DA learning rule",
        description="Synaptic plasticity in the BG input, like the Covariance learning rule with an additional dopamine modulation which depends on the dopamine receptor type ($DA_{type}(D1) = 1, DA_{type}(D2) = -1$).",
    )

    ############################## pallidal synapse (exc) ##################################
    # STNGPi
    DA_excitatory = Synapse(
        parameters="""
            tau=50.0 : projection
            tau_alpha=1.0 : projection
            tau_trace=60.0 : projection
            regularization_threshold=2.6 : projection
            K_burst = 1.0 : projection
            K_dip = 0.4 : projection
            DA_type= 1 : projection
            threshold_pre=0.0 : projection
            threshold_post= -0.15 : projection
            trace_pos_factor = 1.0 : projection
            gamma = 0.45 : projection
        """,
        psp="w * pre.r",
        equations="""
            tau_alpha * dalpha/dt + alpha = pos(post.mp - regularization_threshold)
            dopa_sum = 2.0 * (post.sum(dopa) - baseline_dopa)

            a = mean(post.r) - min(post.r) - gamma : postsynaptic
            post_thresh =   if (-a<threshold_post):
                                -a
                            else:
                                threshold_post : postsynaptic

            trace = pos(pre.r - mean(pre.r) - threshold_pre) * (post.r - mean(post.r) - post_thresh)
            aux = if (trace<0.0): 1 else: 0
            dopa_mod =  if (dopa_sum>0):
                            K_burst * dopa_sum * ((1-trace_pos_factor)*aux+trace_pos_factor)
                        else:
                            K_dip * dopa_sum * aux
            delta = dopa_mod * trace - alpha * pos(trace)
            tau * dw/dt = delta : min=0
        """,
        name="STN Output learning rule",
        description="Synaptic plasticity in the STN output, similar to the Covariance learning rule with an additional dopamine modulation.",
    )

    ############################## pallidal synapse (inh) ##############################
    # StrD1GPi, StrD2GPe
    DA_inhibitory = Synapse(
        parameters="""
            tau=300.0 : projection
            tau_alpha=1.0 : projection
            tau_trace=60.0 : projection
            regularization_threshold=1.0 : projection
            K_burst = 1.0 : projection
            K_dip = 0.4 : projection
            DA_type= 1 : projection
            threshold_pre=0.0 : projection
            threshold_post=0.15 : projection
            trace_neg_factor = 1.0 : projection
            gamma = 0.45 : projection
        """,
        psp="w * pre.r",
        equations="""
            tau_alpha * dalpha/dt + alpha = pos(-post.mp - regularization_threshold)
            dopa_sum = 2.0 * (post.sum(dopa) - baseline_dopa)

            a = mean(post.r) - min(post.r) - gamma : postsynaptic
            post_thresh =   if a>threshold_post:
                                if DA_type>0:
                                    a
                                else:
                                    threshold_post : postsynaptic
                            else:
                                threshold_post : postsynaptic

            trace =     if (DA_type>0):
                            pos(pre.r - mean(pre.r) - threshold_pre) * (mean(post.r) - post.r  - post_thresh)
                        else:
                            pos(pre.r - mean(pre.r) - threshold_pre) * (max(post.r) - post.r  - post_thresh)

            aux = if (trace>0): 1 else: 0

            dopa_mod =  if (DA_type*dopa_sum>0):
                            DA_type*K_burst*dopa_sum * ((1-trace_neg_factor)*aux+trace_neg_factor)
                        else:
                            aux*DA_type*K_dip*dopa_sum

            tau * dw/dt = dopa_mod * trace - alpha * pos(trace) : min=0
        """,
        name="Str Output learning rule",
        description="Synaptic plasticity in the Str output, similar to the Covariance learning (here inverse effect of post-activity) rule with an additional dopamine modulation which depends on the dopamine receptor type ($DA_{type}(D1) = 1, DA_{type}(D2) = -1$).",
    )

    RateCodedSynapse = Synapse(
        parameters = """
            tau = 150.0
            K_burst = 1.0
            K_dip = 0.4
            threshold_pre = 0.05
        """,
        equations = """
            dApre/dt = -Apre / tau : event-driven
            on_pre = { Apre += 1.0 }
            w += K_burst * Apre * reward - K_dip * Apre * (1 - reward)
            w = clip(w, 0.0, 1.0)
        """,
        psp = "w * pre.r"
    )

    ########################## dopamin synapse ##################################
    # StrD1SNc
    DAPrediction = Synapse(
        parameters="""
            tau = 300.0 : projection
        """,
        psp="w * pre.r",
        equations="""
            aux = if (post.sum(exc)>0): 1.0 else: 3.0  : postsynaptic
            tau*dw/dt = aux * (post.r - baseline_dopa) * pos(pre.r - mean(pre.r)) : min=0
        """,
        name="Reward Prediction learning rule",
        description="Simple synaptic plasticity based on covariance.",
    )
    DAPostCovarianceAdaptiveThreshold = Synapse(
    parameters="""
        tau         = 70.0  : projection
        tau_alpha   = 1.0   : projection
        tau_trace   = 60.0  : projection
        tau_thresh  = 500.0 : projection
        regularization_threshold = 1.0 : projection
        K_burst     = 1.0   : projection
        K_dip       = 0.4   : projection
        DA_type     = -1    : projection
        threshold_post = 0.15 : projection
    """,
    psp="w * pre.r",
    equations="""
        tau_alpha * dalpha/dt + alpha = pos(post.mp - regularization_threshold)
        dopa_sum = 2.0 * (post.sum(dopa) - baseline_dopa)

        # محاسبه ترم فعال‌سازی پس‌سیناپسی
        post_term = pos(post.r - mean(post.r) - threshold_post)

        trace = post_term * pre.r

        dopa_mod = if (DA_type * dopa_sum > 0):
                       DA_type * K_burst * dopa_sum
                   else:
                       if (trace > 0.0) and (w > 0.0):
                           DA_type * K_dip * dopa_sum
                       else:
                           0

        # محاسبه تغییرات وزن
        tau * dw/dt = dopa_mod * trace - alpha * post_term * post_term : min=0, max=0.6

        # معادله دیفرانسیل برای یادگیری و به‌روزرسانی آستانه پیش‌سیناپسی
        # مقدار اولیه (init=0.03) جایگزین مقدار پارامتر قبلی شده است
        tau_thresh * dthreshold_pre/dt = dopa_sum * pos(pre.r - threshold_pre) : min=0.0, max=0.5, init=0.03
    """,
    name="DA Covariance with Adaptive Threshold",
    )

    ##############################################################################
    ################################# population #################################
    ##############################################################################

    # IT Input
    IT = Population(name="IT", geometry=params["dim_IT"], neuron=InputNeuron)
    IT.B = params["baseline_IT"]
    IT.phi = params["noise_IT"]

    # Reward Input
    PPN = Population(name="PPN", geometry=params["dim_PPN"], neuron=InputNeuron)
    PPN.tau = 1.0

    # PFC
    PFC = Population(name="PFC", geometry=params["dim_PFC"], neuron=LinearNeuron)
    PFC.phi = params["noise_PFC"]
    PFC.B = params["baseline_PFC"]  # pfc_base

    MTL = Population(name="MTL", geometry=params["dim_MTL"], neuron=MemoryNeuron)
    MTL.B = params["baseline_MTL"]
    MTL.phi = params["noise_MTL"]
    MTL.tau=2000



    # SNc
    SNc = Population(name="SNc", geometry=params["dim_SNc"], neuron=DopamineNeuron)
    SNc.B = params["baseline_SNc"]

    # Striatum direct pathway
    StrD1 = Population(name="StrD1", geometry=params["dim_STR"], neuron=LinearNeuron)
    StrD1.phi = params["noise_Str"]
    StrD1.B = params["baseline_StrD1"]

    # Striatum indirect pathway
    StrD2 = Population(name="StrD2", geometry=params["dim_STR"], neuron=LinearNeuron)
    StrD2.phi = params["noise_Str"]
    StrD2.B = params["baseline_StrD2"]

    # Striatum feedback pathway
    StrThal = Population(
        name="StrThal", geometry=params["dim_StrThal"], neuron=LinearNeuron
    )
    StrThal.phi = params["noise_StrThal"]
    StrThal.B = params["baseline_StrThal"]

    # GPi
    GPi = Population(name="GPi", geometry=params["dim_GPi"], neuron=LinearNeuron)
    GPi.phi = params["noise_GPi"]
    GPi.B = params["baseline_GPi"]

    # STN
    STN = Population(name="STN", geometry=params["dim_STN"], neuron=LinearNeuron)
    STN.phi = params["noise_STN"]
    STN.B = params["baseline_STN"]

    # GPe
    GPe = Population(name="GPe", geometry=params["dim_GPe"], neuron=LinearNeuron)
    GPe.phi = params["noise_GPe"]
    GPe.B = params["baseline_GPe"]

    # Thal
    Thal = Population(name="Thal", geometry=params["dim_Thal"], neuron=LinearNeuron)
    Thal.phi = params["noise_Thal"]
    Thal.B = params["baseline_Thal"]

    ##############################################################################
    ################################# projektion #################################
    ##############################################################################

    ################################# Input ######################################
    # shortcut IT -> Thalamus (plastic or fixed))
    if shortcut == False:
        ITThal = Projection(
            pre=IT, post=Thal, target="exc", synapse=StandardSynapse, name="ITThal"
        )
        ITThal.connect_all_to_all(
            weights=params["ITThal.connect_all_to_all"]
        )  # Normal(0.3,0.1) )

    if shortcut == True:
        ITThal = Projection(
            pre=IT, post=Thal, target="exc", synapse=PostCovariance, name="ITThal"
        )
        ITThal.connect_all_to_all(weights=params["ITThal.connect_all_to_all"])
        ITThal.tau = params["ITThal_tau"]
        ITThal.tau_alpha = params["ITThal_tau_alpha"]
        ITThal.regularization_threshold = params["ITThal_regularization_threshold"]
        ITThal.threshold_post = params["ITThal_threshold_post"]
        ITThal.threshold_pre = params["ITThal_threshold_pre"]
        ITThal.alpha_factor = params["ITThal_alpha_factor"]

    ITMTL= Projection(
        pre=IT,
        post=MTL,
        target="exc",
        synapse=DAPostCovarianceNoThreshold,
        name="ITMTL",
    )
    ITMTL.connect_one_to_one(weights=Uniform (0.15, 0.25))  # Normal(0.15,0.15))
    ITMTL.DA_type = 1
    ITMTL.regularization_threshold = 1.5
    ITMTL.tau = 10
    ITMTL.K_burst = 1.3
    ITMTL.K_dip = 0.2
    # اتصال به استریاتوم با قابلیت یادگیری دوپامینی
    MTLStrD1 = Projection(        pre=MTL,        post=StrD1,
        target="exc",
        synapse=DAPostCovarianceNoThreshold, # تغییر به سیناپس یادگیرنده
        name="MTLStrD1"
    )
    MTLStrD1.connect_all_to_all(weights=Uniform(0.1, 0.3))
    MTLStrD1.DA_type = 1 # حساس به پاداش    # فعال کردن حلقه PFC برای پایداری حافظه (اختیاری اما توصیه شده)
    MTLPFC = Projection(pre=MTL, post=PFC, target="exc", synapse=StandardSynapse)
    MTLPFC.connect_all_to_all(weights=0.04)

    PFCMTL = Projection(pre=PFC, post=MTL, target="exc", synapse=StandardSynapse)
    PFCMTL.connect_all_to_all(weights=0.04)
    MTLStrD2 = Projection(        pre=MTL,        post=StrD2,        target="exc",
        synapse=DAPostCovarianceNoThreshold, # تغییر به سیناپس یادگیرنده
        name="MTLStrD2"
    )
    MTLStrD2.connect_all_to_all(weights=Uniform(0.1, 0.3))
    MTLStrD2.DA_type = -1 # حساس به پاداش    # فعال کردن حلقه PFC برای پایداری حافظه (اختیاری اما توصیه شده)

    PFCMTL = Projection(pre=PFC, post=MTL, target="exc", synapse=StandardSynapse)
    PFCMTL.connect_all_to_all(weights=0.04)

    #MTLPFC = Projection(
        #pre=MTL, post=PFC, target="exc", synapse=StandardSynapse, name="MTLPFC"
    #)
    #MTLPFC.connect_all_to_all(weights=0.05)

    #PFCMTL = Projection(
        #pre=PFC, post=MTL, target="exc", synapse=StandardSynapse, name="PFCMTL"
    #)
    #PFCMTL.connect_all_to_all(weights=0.05)
    #MTLStrD1 = Projection(
        #pre=MTL, post=StrD1, target="exc", synapse=StandardSynapse, name="MTLStrD1"
    #)
     #پیشنهاد: از وزن کم شروع کن تا سیستم نوسانی نشه
    #MTLStrD1.connect_all_to_all(weights=0.02)


    ITStrD1 = Projection(
        pre=IT,
        post=StrD1,
        target="exc",
        synapse=DAPostCovarianceNoThreshold,
        name="ITStrD1",
    )
    ITStrD1.connect_all_to_all(weights=Uniform (0.1, 0.3))  # Normal(0.15,0.15))
    ITStrD1.DA_type = 1
    ITStrD1.regularization_threshold = params["ITStrD1_regularization_threshold"]
    ITStrD1.tau = params["ITStrD1_tau"]
    ITStrD1.K_burst = params["ITStrD1_K_burst"]
    ITStrD1.K_dip = params["ITStrD1_K_dip"]

    ITStrD2 = Projection(
        pre=IT,
        post=StrD2,
        target="exc",
        synapse=DAPostCovarianceNoThreshold,
        name="ITStrD2",
    )
    ITStrD2.connect_all_to_all(weights=Uniform (0.1, 0.3))  # Normal(0.15,0.15))
    ITStrD2.DA_type = -1
    ITStrD2.regularization_threshold = params["ITStrD2_regularization_threshold"]
    ITStrD2.tau = params["ITStrD2_tau"]
    ITStrD2.K_burst = params["ITStrD2_K_burst"]
    ITStrD2.K_dip = params["ITStrD2_K_dip"]

    #ITSTN = Projection(
       # pre=IT,
       # post=STN,
       # target="exc",
       # synapse=DAPostCovarianceNoThreshold,
      #  name="ITSTN",
  #  )
   # ITSTN.connect_all_to_all(weights=Uniform(0.1, 0.3))  # Normal(0.15,0.15))
   # ITSTN.DA_type = -1
   # ITSTN.regularization_threshold = params["ITSTN_regularization_threshold"]
    #ITSTN.tau = params["ITSTN_tau"]
   # ITSTN.K_burst = params["ITSTN_K_burst"]
   # ITSTN.K_dip = params["ITSTN_K_dip"]

    ITSTN = Projection(
    pre=IT,
    post=STN,
    target="exc",
    synapse=DAPostCovarianceAdaptiveThreshold,  # ← synapse جدید
    name="ITSTN",
    )
    ITSTN.connect_all_to_all(weights=Uniform(0.1, 0.3))
    ITSTN.DA_type = -1
    ITSTN.tau_thresh = 300.0  # هر چی بزرگتر، کندتر یاد می‌گیره

    ########################## output ##################################

    GPiThal = Projection(pre=GPi, post=Thal, target="inh", synapse=StandardSynapse, name="GPiThal")
    GPiThal.connect_one_to_one(weights=params["GPiThal.connect_one_to_one"])

    ThalPFC = Projection(
        pre=Thal, post=PFC, target="exc", synapse=StandardSynapse, name="ThalPFC"
    )
    ThalPFC.connect_one_to_one(weights=params["ThalPFC.connect_one_to_one"])  # init

    ########################## dopamin ##################################

    PPNSNc = Projection(
        pre=PPN, post=SNc, target="exc", synapse=StandardSynapse, name="PPNSNc"
    )
    PPNSNc.connect_all_to_all(weights=1.0)

    StrD1SNc = Projection(
        pre=StrD1, post=SNc, target="inh", synapse=DAPrediction, name="StrD1SNc"
    )
    StrD1SNc.connect_all_to_all(weights=0.05)  # 0.5

    SNcStrD1 = Projection(
        pre=SNc, post=StrD1, target="dopa", synapse=StandardSynapse, name="SNcStrD1"
    )
    SNcStrD1.connect_all_to_all(weights=1.0)  # 1.0

    SNcStrD2 = Projection(
        pre=SNc, post=StrD2, target="dopa", synapse=StandardSynapse, name="SNcStrD2"
    )
    SNcStrD2.connect_all_to_all(weights=1.0)

    SNcGPi = Projection(
        pre=SNc, post=GPi, target="dopa", synapse=StandardSynapse, name="SNcGPi"
    )
    SNcGPi.connect_all_to_all(weights=1.0)

    SNcSTN = Projection(
        pre=SNc, post=STN, target="dopa", synapse=StandardSynapse, name="SNcSTN"
    )
    SNcSTN.connect_all_to_all(weights=1.0)

    SNcGPe = Projection(
        pre=SNc, post=GPe, target="dopa", synapse=StandardSynapse, name="SNcGPe"
    )
    SNcGPe.connect_all_to_all(weights=1.0)

    SNcMTL = Projection(
        pre=SNc, post=MTL, target="dopa", synapse=StandardSynapse, name="SNcMTL"
    )
    SNcMTL.connect_all_to_all(weights=1.0)

    ########################## Inner BG ######################################
    # --- StrD1 -> GPi (inhibitory, DA-modulated) ---

    StrD1GPi = Projection(
        pre=StrD1, post=GPi, target="inh", synapse=DA_inhibitory, name="StrD1GPi"
    )
    StrD1GPi.connect_all_to_all(weights=Uniform(0.05, 0.08))  # Normal(0.025,0.025))
    StrD1GPi.DA_type = 1
    StrD1GPi.threshold_post = params["StrD1GPi_threshold_post"]
    StrD1GPi.trace_neg_factor = params["StrD1GPi_trace_neg_factor"]
    StrD1GPi.regularization_threshold = params["StrD1GPi_regularization_threshold"]
    StrD1GPi.tau = params["StrD1GPi_tau"]
    StrD1GPi.K_burst = params["StrD1GPi_K_burst"]
    StrD1GPi.K_dip = params["StrD1GPi_K_dip"]

    STNGPi = Projection(
        pre=STN, post=GPi, target="exc", synapse=DA_excitatory, name="STNGPi"
    )
    STNGPi.connect_all_to_all(weights=Uniform(0.1, 0.25))  # Normal(0.025,0.025))
    STNGPi.DA_type= 1
    STNGPi.threshold_post = params["STNGPi_threshold_post"]
    STNGPi.trace_neg_factor = params["STNGPi_trace_neg_factor"]
    STNGPi.regularization_threshold = params["STNGPi_regularization_threshold"]
    STNGPi.tau = params["STNGPi_tau"]
    STNGPi.K_burst = params["STNGPi_K_burst"]
    STNGPi.K_dip = params["STNGPi_K_dip"]

    StrD2GPe = Projection(
        pre=StrD2, post=GPe, target="inh", synapse=DA_inhibitory, name="StrD2GPe"
    )
    StrD2GPe.connect_all_to_all(weights=Uniform(0.05, 0.08))
      # Normal(0.025,0.025))
    StrD2GPe.DA_type = -1
    StrD2GPe.threshold_post = params["StrD2GPe_threshold_post"]
    StrD2GPe.trace_neg_factor = params["StrD2GPe_trace_neg_factor"]
    StrD2GPe.regularization_threshold = params["StrD2GPe_regularization_threshold"]
    StrD2GPe.tau = params["StrD2GPe_tau"]
    StrD2GPe.K_burst = params["StrD2GPe_K_burst"]
    StrD2GPe.K_dip = params["StrD2GPe_K_dip"]

    GPeGPi = Projection(pre=GPe, post=GPi, target="inh", synapse=StandardSynapse)
    GPeGPi.connect_one_to_one(weights=params["GPeGPi.connect_one_to_one"])

    GPeSTN = Projection(pre=GPe, post=STN, target="inh", synapse=StandardSynapse)
    GPeSTN.connect_all_to_all(weights=params["GPeSTN.connect_all_to_all"])

    STNGPe = Projection(pre=STN, post=GPe, target="exc", synapse=StandardSynapse)
    STNGPe.connect_all_to_all(weights=params["STNGPe.connect_all_to_all"])

    ####################### lateral connection ##################################

    #MTLMTL = Projection(
        #pre=MTL, post=MTL, target="inh", synapse=StandardSynapse, name='MTLMTL'
    #)
    #MTLMTL.connect_all_to_all(weights=0.008)
  #  MTLMTL = Projection(pre=MTL, post=MTL, target='exc')
   # MTLMTL.connect_dog(amp_pos=0.01, sigma_pos=4.0, amp_neg=0.1, sigma_neg=3.0)
    MTL_inh = Projection(pre=MTL, post=MTL, target='inh')
    MTL_inh.connect_all_to_all(weights=0.2) # یک مهار ضعیف و پخش شده

# ۲. تحریک موضعی (Local Excitation) با DoG - معادل فلش‌های مشکی
# در اینجا باید بخش مثبت (تحریک) قوی‌تر و متمرکزتر باشد
   # MTL_exc = Projection(pre=MTL, post=MTL, target='exc')
    #MTL_exc.connect_dog(
     # amp_pos=0.1,   # تحریک قوی در مرکز
     # sigma_pos=1.0,  # فقط همسایگان خیلی نزدیک (متمرکز)
     # amp_neg=0.05,   # مهار جانبی ضعیف
    #  sigma_neg=2.0   # در محدوده وسیع‌تر
  #  )


    StrD1StrD1 = Projection(
        pre=StrD1, post=StrD1, target="inh", synapse=StandardSynapse
    )
    StrD1StrD1.connect_all_to_all(weights=params["StrD1StrD1.connect_all_to_all"])

    STNSTN = Projection(pre=STN, post=STN, target="inh", synapse=StandardSynapse)
    STNSTN.connect_all_to_all(weights=params["STNSTN.connect_all_to_all"])

    StrD2StrD2 = Projection(
        pre=StrD2, post=StrD2, target="inh", synapse=StandardSynapse
    )
    StrD2StrD2.connect_all_to_all(weights=params["StrD2StrD2.connect_all_to_all"])

    StrThalStrThal = Projection(
        pre=StrThal, post=StrThal, target="inh", synapse=StandardSynapse
    )
    StrThalStrThal.connect_all_to_all(
        weights=params["StrThalStrThal.connect_all_to_all"]
    )

    GPiGPi = Projection(pre=GPi, post=GPi, target="inh", synapse=StandardSynapse, name="GPiGPi")
    GPiGPi.connect_all_to_all(weights=params["GPiGPi.connect_all_to_all"])
    GPeGPe = Projection(pre=GPe, post=GPe, target="inh", synapse=StandardSynapse, name="GPeGPe")
    GPeGPe.connect_all_to_all(weights=params["GPeGPe.connect_all_to_all"])

    ########################### Thalamus Feedback ################################

    ThalStrThal = Projection(
        pre=Thal, post=StrThal, target="exc", synapse=StandardSynapse
    )
    ThalStrThal.connect_one_to_one(weights=params["ThalStrThal.connect_one_to_one"])

    StrThalGPe = Projection(
        pre=StrThal, post=GPe, target="inh", synapse=StandardSynapse
    )
    StrThalGPe.connect_one_to_one(weights=params["StrThalGPe.connect_one_to_one"])

    StrThalGPi = Projection(
        pre=StrThal, post=GPi, target="inh", synapse=StandardSynapse
    )
    StrThalGPi.connect_one_to_one(weights=params["StrThalGPi.connect_one_to_one"])

    ##############################################################################
    ############################### DBS Stimulator ###############################
    ##############################################################################

    if dbs_param_state == 0:
        # dbs_state = 0 -> DBS-OFF
        if dbs_state == 0:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=0.0,
                dbs_pulse_frequency_Hz=0.0,
                dbs_pulse_width_us=0.0,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 1 -> supress lokal neurons
        if dbs_state == 1:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["1_suppression"],
                dbs_pulse_frequency_Hz=params["1_frequence"],
                dbs_pulse_width_us=params["1_pulse_width"],
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 2 -> activate efferent axons
        if dbs_state == 2:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["2_suppression"],
                dbs_pulse_frequency_Hz=params["2_frequence"],
                dbs_pulse_width_us=params["2_pulse_width"],
                axon_rate_amp=params["2_axon_rate_amplitude"],
                orthodromic=True,
                efferents=True,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 3 -> activate afferent axons
        if dbs_state == 3:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["3_suppression"],
                dbs_pulse_frequency_Hz=params["3_frequence"],
                dbs_pulse_width_us=params["3_pulse_width"],
                axon_rate_amp=params["3_axon_rate_amplitude"],
                orthodromic=True,
                afferents=True,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 4 -> activate passing neurons
        if dbs_state == 4:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["4_suppression"],
                dbs_pulse_frequency_Hz=params["4_frequence"],
                dbs_pulse_width_us=params["4_pulse_width"],
                orthodromic=True,
                passing_fibres=True,
                passing_fibres_list=[
                    GPeSTN,
                ],
                axon_rate_amp=params["4_axon_rate_amplitude"],
                passing_fibres_strength=params["4_fibre_strength"],
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 5 -> dbs-comb
        if dbs_state == 5:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["5_suppression"],
                dbs_pulse_frequency_Hz=params["5_frequence"],
                dbs_pulse_width_us=params["5_pulse_width"],
                axon_rate_amp=params["5_axon_rate_amplitude"],
                orthodromic=True,
                efferents=True,
                afferents=False,
                passing_fibres=True,
                passing_fibres_list=[
                    GPeSTN,
                ],
                passing_fibres_strength=params["5_fibre_strength"],
                seed=seed,
                auto_implement=True,
            )
    else:
        # dbs_state = 0 -> DBS-OFF
        if dbs_state == 0:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=0.0,
                dbs_pulse_frequency_Hz=0.0,
                dbs_pulse_width_us=0.0,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 1 -> supress lokal neurons
        if dbs_state == 1:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=parameter,
                dbs_pulse_frequency_Hz=params["1_frequence"],
                dbs_pulse_width_us=params["1_pulse_width"],
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 2 -> activate efferent axons
        if dbs_state == 2:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["2_suppression"],
                dbs_pulse_frequency_Hz=params["2_frequence"],
                dbs_pulse_width_us=params["2_pulse_width"],
                axon_rate_amp=parameter,
                orthodromic=True,
                efferents=True,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 3 -> activate afferent axons
        if dbs_state == 3:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["3_suppression"],
                dbs_pulse_frequency_Hz=params["3_frequence"],
                dbs_pulse_width_us=params["3_pulse_width"],
                axon_rate_amp=parameter,
                orthodromic=True,
                afferents=True,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 4 -> activate passing neurons
        if dbs_state == 4:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["4_suppression"],
                dbs_pulse_frequency_Hz=params["4_frequence"],
                dbs_pulse_width_us=params["4_pulse_width"],
                orthodromic=True,
                passing_fibres=True,
                passing_fibres_list=[
                    GPeSTN,
                ],
                axon_rate_amp=params["4_axon_rate_amplitude"],
                passing_fibres_strength=parameter,
                seed=seed,
                auto_implement=True,
            )

        # dbs_state = 5 -> dbs-comb
        if dbs_state == 5:
            dbs = DBSstimulator(
                stimulated_population=GPi,
                dbs_depolarization=params["5_suppression"],
                dbs_pulse_frequency_Hz=params["5_frequence"],
                dbs_pulse_width_us=params["5_pulse_width"],
                axon_rate_amp=params["5_axon_rate_amplitude"],
                orthodromic=True,
                efferents=True,
                afferents=False,
                passing_fibres=True,
                passing_fibres_list=[
                    GPeSTN,
                ],
                passing_fibres_strength=params["5_fibre_strength"],
                seed=seed,
                auto_implement=True,
            )

    ########################### set pointer on population #############################
    (
        IT,
        PFC,
        StrD1,
        StrD2,
        STN,
        GPe,
        GPi,
        Thal,
        SNc,
        StrThal,
        PPN,
        PFC,
        ITStrD1,
        ITStrD2,
        #MTLStrD1,
        ITSTN,
        StrD1GPi,
        StrD2GPe,
        STNGPi,
        StrD1SNc,
        ITThal,
        ThalPFC,
    ) = dbs.update_pointers(
        [
            IT,
            PFC,
            StrD1,
            StrD2,
            STN,
            GPe,
            GPi,
            Thal,
            SNc,
            StrThal,
            PPN,
            PFC,
            ITStrD1,
            ITStrD2,
            #MTLStrD1,
            ITSTN,
            StrD1GPi,
            StrD2GPe,
            STNGPi,
            StrD1SNc,
            ITThal,
            ThalPFC,
        ]
    )

    ########################### compile network #############################
    compile_in_folder(folder_name=ann_compile_str)

    populations = [
        IT,
        PFC,
        StrD1,
        StrD2,
        STN,
        GPe,
        GPi,
        Thal,
        SNc,
        StrThal,
        PPN,
        PFC,
        MTL,
        #ITMTL,
        ITStrD1,
        ITStrD2,
        #MTLStrD1,
        ITSTN,
        StrD1GPi,
        StrD2GPe,
        STNGPi,
        StrD1SNc,
        ITThal,
        ThalPFC,
        dbs,
    ]

    return populations


##############################################################################
################################# Monitoring #################################
##############################################################################

########################### record firing rates ##############################


class BGMonitor(object):
    def __init__(self, populations):
        self.populations = populations
        self.monitors = []
        for pop in populations:
            self.monitors.append(Monitor(pop, "r", start=False))

    def start(self):
        for monitor in self.monitors:
            monitor.start()

    def stop(self):
        for monitor in self.monitors:
            monitor.pause()

    def get(self):
        res = {}
        for monitor in self.monitors:
            res[monitor.object.name] = monitor.get("r")
        return res


############################### record weigths ###############################


def get_plastic_weights(data: dict):
    """
    Everytime called it stores the weights of the plastic projections in a dictionary
    (keys = projection names, values = arrays with projection weights for each call).
    The arrays have shape (n,) + projection.w.shape. Only works if projection.w can be
    transformed into an array (e.g., for one-to-one or all-to-all connections).

    Args:
        data (dict):
            Either empty dictionary or dictionary already containing arrays of weights
            of the plastic projections.

    Returns:
        data (dict):
            Dictionary with updated arrays.
    """
    # get the plastic projections
    plastic_proj_list: list[Projection] = [
        proj
        for proj in projections()
        if "dw" in re.sub(r"\s+", "", proj.synapse_type.equations)
    ]

    # create or append the dictionary entries for the projection weights
    for proj in plastic_proj_list:
        w = np.array(proj.w)
        w = np.expand_dims(w, axis=0)
        key = proj.name

        if key in data:
            # Concatenate along the first axis
            data[key] = np.concatenate((data[key], w), axis=0)
        else:
            # Initialize with the first array
            data[key] = w

    return data


def extract_data(
    SNc,
    ITStrD1,
    ITStrD2,
    ITSTN,
    StrD1GPi,
    StrD2GPe,
    STNGPi,
    StrD1SNc,
    ITThal,
    ThalPFC,
    shortcut,
):

    # 0
    IT_StrD1 = np.array([dendrite.w for dendrite in ITStrD1])
    IT_StrD1[IT_StrD1 < 0] = 0.0  # Alles kleiner 0 wird durch 0 ersetzt
    IT_StrD1 = np.reshape(
        IT_StrD1, (params["dim_IT"] * params["dim_STR"], 1)
    )  # Array in 8x1 Dimension wandeln
    IT_StrD1 = IT_StrD1.tolist()  # Array zu Liste aus Arrays wandeln
    # 1
    IT_StrD2 = np.array([dendrite.w for dendrite in ITStrD2])
    IT_StrD2[IT_StrD2 < 0] = 0.0
    IT_StrD2 = np.reshape(IT_StrD2, (params["dim_IT"] * params["dim_STR"], 1))
    IT_StrD2 = IT_StrD2.tolist()
    # 2
    IT_STN = np.array([dendrite.w for dendrite in ITSTN])
    IT_STN[IT_STN < 0] = 0.0
    IT_STN = np.reshape(IT_STN, (params["dim_IT"] * params["dim_STN"], 1))
    IT_STN = IT_STN.tolist()
    # 3
    StrD1_GPi = np.array([dendrite.w for dendrite in StrD1GPi])
    StrD1_GPi[StrD1_GPi < 0] = 0.0
    StrD1_GPi = np.reshape(StrD1_GPi, (params["dim_GPi"] * params["dim_STR"], 1))
    StrD1_GPi = StrD1_GPi.tolist()
    # 4
    StrD2_GPe = np.array([dendrite.w for dendrite in StrD2GPe])
    StrD2_GPe[StrD2_GPe < 0] = 0.0
    StrD2_GPe = np.reshape(StrD2_GPe, (params["dim_GPe"] * params["dim_STR"], 1))
    StrD2_GPe = StrD2_GPe.tolist()
    # 5
    STN_GPi = np.array([dendrite.w for dendrite in STNGPi])
    STN_GPi[STN_GPi < 0] = 0.0
    STN_GPi = np.reshape(STN_GPi, (params["dim_GPi"] * params["dim_STN"], 1))
    STN_GPi = STN_GPi.tolist()
    # 6
    StrD1_SNc = np.array([dendrite.w for dendrite in StrD1SNc])
    StrD1_SNc[StrD1_SNc < 0] = 0.0
    StrD1_SNc = np.reshape(StrD1_SNc, (params["dim_STR"], 1))
    StrD1_SNc = StrD1_SNc.tolist()
    # 7
    IT_Thal = np.array([dendrite.w for dendrite in ITThal])
    IT_Thal[IT_Thal < 0] = 0.0
    if shortcut == False:
        IT_Thal = np.reshape(IT_Thal, (params["dim_IT"], 1))
    if shortcut == True:
        IT_Thal = np.reshape(IT_Thal, (params["dim_STR"], 1))
    IT_Thal = IT_Thal.tolist()
    # 8
    Thal_PFC = np.array([dendrite.w for dendrite in ThalPFC])
    Thal_PFC[Thal_PFC < 0] = 0.0
    Thal_PFC = np.reshape(Thal_PFC, (params["dim_Thal"], 1))
    Thal_PFC = Thal_PFC.tolist()
    # 9 Dopamine level
    DA = SNc.r
    DA = DA.tolist()

    gewichte = {
        "IT_StrD1": IT_StrD1,
        "IT_StrD2": IT_StrD2,
        "IT_STN": IT_STN,
        "StrD1_GPi": StrD1_GPi,
        "StrD2_GPe": StrD2_GPe,
        "STN_GPi": STN_GPi,
        "StrD1_SNc": StrD1_SNc,
        "IT_Thal": IT_Thal,
        "Thal_PFC": Thal_PFC,
        "DA": DA,
    }
    return gewichte


########################### record threshold ##################################
```

```python

# ==============================================================================
# 2. Functional Validation: Direct Pathway Circuitry
# ==============================================================================
# This procedure performs a 'smoke test' to verify the integrity of the direct
# pathway. It assesses the signal propagation from the Inferior Temporal (IT)
# cortex through the striatal (StrD1) inhibition, GPi disinhibition,
# and thalamo-cortical (Thal-PFC) activation.

def calculate_mean_rate(population):
    """Safely calculates the mean firing rate of an ANNarchy population."""
    try:
        arr = np.array(population.r, dtype=float).ravel()
        return float(arr.mean()) if arr.size else float('nan')
    except Exception as e:
        print(f"Error calculating mean rate: {e}")
        return float('nan')

# Step 1: Capture baseline neural activity (pre-stimulus)
ann.simulate(100.0)
baseline_metrics = {
    'StrD1': calculate_mean_rate(StrD1),
    'STN'  : calculate_mean_rate(STN),
    'GPi'  : calculate_mean_rate(GPi),
    'Thal' : calculate_mean_rate(Thal),
    'PFC'  : calculate_mean_rate(PFC),
}

# Step 2: Apply transient IT cortical pulse to stimulate the direct pathway
IT.B = 1.0
ann.simulate(150.0)
post_metrics = {
    'StrD1': calculate_mean_rate(StrD1),
    'STN'  : calculate_mean_rate(STN),
    'GPi'  : calculate_mean_rate(GPi),
    'Thal' : calculate_mean_rate(Thal),
    'PFC'  : calculate_mean_rate(PFC),
}

# Step 3: Restore system to baseline
IT.B = 0.0
ann.simulate(100.0)

# Report results for dissertation documentation
print(f"--- Smoke Test Results ---")
print(f"Baseline: GPi={baseline_metrics['GPi']:.3f}, Thal={baseline_metrics['Thal']:.3f}, PFC={baseline_metrics['PFC']:.3f}")
print(f"Post-Pulse: StrD1={post_metrics['StrD1']:.3f}, GPi={post_metrics['GPi']:.3f}, Thal={post_metrics['Thal']:.3f}, PFC={post_metrics['PFC']:.3f}")

# Verification of pathway logic
logic_check = {
    "Direct Path (GPi inhibition)": post_metrics['GPi'] < baseline_metrics['GPi'],
    "Thalamic Disinhibition": post_metrics['Thal'] > baseline_metrics['Thal'],
    "Cortical Drive (PFC excitation)": post_metrics['PFC'] > baseline_metrics['PFC']
}

for test, passed in logic_check.items():
    print(f"{test:<35}: {'PASS ✅' if passed else 'FAIL ❌'}")
```

```python


def get_fish(fish_list, feature_name_list):
    """
    return a dictionary with features and feature values

    Args:
        fish_list: list
            list with all possible fishes consisting of 4 feature values
        feature_name_list: list
            list containing 4 strings, the names of the 4 features (the order corresponds to the order in fish_list)
    """
    idx1 =random.randint(0, 1)

    idx2 =random.randint( 0,5)

    input_sacc = fish_list[idx1][idx2]

    fish = dict(zip(feature_name_list, input_sacc))
    return fish
def get_true_category(fish):
    important_feature =  "DF"
    important_feature1= "TF"
    important_feature2= "VF"
    """
    returns the of a given fish based on a single feature
    """

    if fish[important_feature]  - fish[important_feature1] < 0.0:
        cat = "A"

    if fish[important_feature] - fish[important_feature1] > 0.0:
        cat = "B"
    if fish[important_feature] - fish[important_feature1] == 0.0:
       # cat = "A"
        if fish[important_feature2]==-0.5:
           cat="A"
        else : cat="B"

    return cat
def select_input(feature_name, fish, IT, feature_name_list, baseline_default):
    """
    activates IT neurons based on the presented feature

    Args:
        feature_name: string
            name of the activated feature
        fish: dictionary
            dictionary containing the feature values for the fish features
        IT: ANNarchy population
            the IT population
        feature_name_list: list
            list containing 4 strings, the names of the 4 features (the order corresponds to the order in fish_list)
        baseline_default: float
            default baseline of IT
    """
    #baseline=B
    # set deafult baseline
    #IT.B = IT.B*0.2

    # feature name i.e. spatial position defines the group of IT neurons
    nr_IT_neurons = len(IT)
    nr_group_neurons = nr_IT_neurons // (len(feature_name_list))
    neuron_group_idx = feature_name_list.index(feature_name)
    neuron_group = IT[
        int(neuron_group_idx * nr_group_neurons): int(
            (neuron_group_idx + 1) * nr_group_neurons
        )
    ]

    # feature value defines which neurons of neuron group get active --> neuron groups are feature-value selective
    feature_value = fish[feature_name]
    activity_center_idx = ((feature_value - (-1)) /
                           (1 - (-1))) * (nr_group_neurons - 1)

    activity = gauss_1D(
        m=activity_center_idx,
        sig=nr_group_neurons / 10,
        size=nr_group_neurons,
        min=neuron_group.B[0],
    )

    # set the activity of the neuron group
    neuron_group.B = activity
def gauss_1D(m, sig, size, min):
    """
    returns a 1D gaussian array

    Args:
        m: float
            center/mean of the gaussian along the array indizes
        sig: float
            standard deviation in array indizes
        size: int
            size of the returned array
        min: float
            minimum value
    """
    return np.clip(np.exp(-((np.arange(size) - m) ** 2) / (2 * sig**2)), min, None)


def get_feature_order(feature_name_list):
    """
    return a list with feature names containing the presentation order of the fish features
    """
    idx_arr = np.arange(len(feature_name_list))
    feature_name_arr = np.array(feature_name_list)
    np.random.shuffle(idx_arr)
    return feature_name_arr[idx_arr].tolist()
```

```python

from collections import deque
import numpy as np
import random
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from collections import defaultdict
from ANNarchy import simulate, reset
from CompNeuroPy import CompNeuroMonitors, get_full_model

try:
    from ANNarchy import dt as _ann_dt
except ImportError:
    _ann_dt = None


class CategorizationExperiment:

    @staticmethod
    def setup_monitors(verbose=True):
        full_model = get_full_model()
        populations = full_model["populations"]
        monitor_config = {pop_name: ["r"] for pop_name in populations}
        mons = CompNeuroMonitors(monitor_config)

        if verbose:
            print(f"Monitors initialized for {len(populations)} populations.")
            print(f"Monitoring configuration: {list(monitor_config.keys())}")

        return mons, populations

    def __init__(
        self,
        mons, IT, PFC, PPN, SNc, StrD1GPi, StrD2GPe,
        fish_list, feature_name_list,
        get_fish, get_true_category, get_feature_order, select_input,
        n_blocks=5,
        n_repeats=2,
        nr_fish_presentations=150,
        trial_duration=210,
        settle_duration=40,
        dopamine_input_duration=290,
        baseline_it=0.0,
        response_threshold=0.7,
        margin_threshold=0.5,
        seed=44,
        dt=None,
        weight_clip_min=0.05,
        weight_clip_max=2.8,
        dopamine_correct=1.0,
        dopamine_incorrect=0.05,
    ):
        self.mons = mons
        self.IT = IT
        self.PFC = PFC
        self.PPN = PPN
        self.SNc = SNc
        self.StrD1GPi = StrD1GPi
        self.StrD2GPe = StrD2GPe

        self.fish_list = fish_list
        self.feature_name_list = feature_name_list
        self.get_fish = get_fish
        self.get_true_category = get_true_category
        self.get_feature_order = get_feature_order
        self.select_input = select_input

        self.n_blocks = n_blocks
        self.n_repeats = n_repeats
        self.nr_fish_presentations = nr_fish_presentations
        self.trial_duration = trial_duration
        self.settle_duration = settle_duration
        self.dopamine_input_duration = dopamine_input_duration
        self.baseline_it = baseline_it

        self.response_threshold = response_threshold
        self.margin_threshold = margin_threshold

        self.weight_clip_min = weight_clip_min
        self.weight_clip_max = weight_clip_max

        self.dopamine_correct = dopamine_correct
        self.dopamine_incorrect = dopamine_incorrect

        self.seed = seed
        np.random.seed(seed)

        if dt is not None:
            self.dt = dt
        elif _ann_dt is not None:
            try:
                self.dt = _ann_dt()
            except Exception:
                self.dt = 1.0
        else:
            self.dt = 1.0

        self.performance_matrix = np.zeros((n_blocks, n_repeats))

        self.trial_log = {
            b: {r: [] for r in range(n_repeats)} for b in range(n_blocks)
        }

        self._chunk_counter = 0

    # ---------------------------------------------------------
    # Single Trial
    # ---------------------------------------------------------
    def run_trial(self):
        simulate(120)

        fish = self.get_fish(self.fish_list, self.feature_name_list)
        category_true = self.get_true_category(fish)
        feature_order = self.get_feature_order(self.feature_name_list)
        #feature_order=['MA','DF','VF','TF']
        category_selected = None
        trial_correct = 0
        decided_at_feature = None
        resp_val_final = None
        margin_final = None

        self.mons.start()

        attempts_used = 0
        for idx, feature_name in enumerate(feature_order, start=1):
            attempts_used = idx

            self.select_input(
                feature_name, fish, self.IT,
                self.feature_name_list, self.baseline_it
            )

            simulate(self.trial_duration + self.settle_duration)

            pfc_act = np.array(self.PFC.r).flatten()
            resp_val = np.max(pfc_act)
            resp_idx = np.argmax(pfc_act)

            sorted_pfc = np.sort(pfc_act)
            margin = sorted_pfc[-1] - sorted_pfc[-2]

            dynamic_thresh = (
                0.7 if (idx == len(feature_order)) else self.response_threshold
            )

            if resp_val >= dynamic_thresh and margin >= self.margin_threshold:
                category_selected = ["A", "B"][resp_idx]
                trial_correct = int(category_selected == category_true)
                decided_at_feature = feature_name
                resp_val_final = resp_val
                margin_final = margin

                if trial_correct:
                    self.PPN.B = 1.0
                    self.SNc.alpha = self.dopamine_correct
                else:
                    self.PPN.B = 0.0
                    self.SNc.alpha = self.dopamine_incorrect

                simulate(self.dopamine_input_duration)

                self.StrD1GPi.w = np.clip(
                    self.StrD1GPi.w,
                    self.weight_clip_min,
                    self.weight_clip_max
                )
                self.StrD2GPe.w = np.clip(
                    self.StrD2GPe.w,
                    self.weight_clip_min,
                    self.weight_clip_max
                )
                break

        self.mons.pause()

        self.IT.B = 0
        self.PPN.B = 0
        self.SNc.alpha = 0
        simulate(120)

        recorded_ms = attempts_used * (self.trial_duration + self.settle_duration)
        if decided_at_feature is not None:
            recorded_ms += self.dopamine_input_duration

        trial_info = {
            "fish": fish,
            "category_true": category_true,
            "category_selected": category_selected,
            "correct": trial_correct,
            "feature_order": feature_order,
            "decided_at_feature": decided_at_feature,
            "response_value": resp_val_final,
            "margin": margin_final,
            "recorded_ms": recorded_ms,
            "n_samples": int(round(recorded_ms / self.dt)),
        }

        return trial_info

    # ---------------------------------------------------------
    # Repeat
    # ---------------------------------------------------------
    def run_repeat(self, block_idx, repeat_idx, print_trials=False):
        self.mons.reset()
        chunk_idx = self._chunk_counter
        self._chunk_counter += 1

        offset = 0
        block_accuracies = deque()

        for trial_idx in range(self.nr_fish_presentations):
            trial_info = self.run_trial()

            trial_info["trial_idx"] = trial_idx
            trial_info["chunk_idx"] = chunk_idx
            trial_info["start_idx"] = offset
            trial_info["end_idx"] = offset + trial_info["n_samples"]
            offset = trial_info["end_idx"]

            self.trial_log[block_idx][repeat_idx].append(trial_info)
            block_accuracies.append(trial_info["correct"])

            if print_trials:
                resp_str = f"{trial_info['response_value']:.3f}" if trial_info['response_value'] is not None else "None"
                margin_str = f"{trial_info['margin']:.3f}" if trial_info['margin'] is not None else "None"
                feature_str = str(trial_info['decided_at_feature']) if trial_info['decided_at_feature'] is not None else "None"

                print(
                    f"    Trial {trial_idx + 1:3d}: "
                    f"correct={trial_info['correct']} | "
                    f"true={trial_info['category_true']} | "
                    f"selected={trial_info['category_selected']} | "
                    f"resp_val={resp_str} | "
                    f"margin={margin_str} | "
                    f"feature={feature_str}"
                )

        return float(np.mean(block_accuracies))

    # ---------------------------------------------------------
    # Full Run
    # ---------------------------------------------------------
    def run(self, verbose=True, print_trials=False):
        np.random.seed(self.seed)

        for block_idx in range(self.n_blocks):
            if verbose:
                print(f"\n===== Block {block_idx + 1}/{self.n_blocks} =====")

            reset()
            simulate(50)

            for repeat in range(self.n_repeats):
                acc = self.run_repeat(block_idx, repeat, print_trials=print_trials)
                self.performance_matrix[block_idx, repeat] = acc
                if verbose:
                    print(f"  Repeat {repeat + 1} accuracy: {acc:.3f}")

        return self.performance_matrix

    # ---------------------------------------------------------
    # Trial Info Helpers
    # ---------------------------------------------------------
    def get_trial_info(self, block_idx, trial_idx, repeat_idx=0):
        try:
            return self.trial_log[block_idx][repeat_idx][trial_idx]
        except (KeyError, IndexError):
            raise ValueError("Invalid block/repeat/trial index.")

    def get_block_info(self, block_idx, repeat_idx=0):
        try:
            return self.trial_log[block_idx][repeat_idx]
        except KeyError:
            raise ValueError("Invalid block/repeat index.")

    def get_trial_recording(self, block_idx, trial_idx, repeat_idx=0):
        info = self.get_trial_info(block_idx, trial_idx, repeat_idx)
        recordings = self.mons.get_recordings()
        full_rec = recordings[info["chunk_idx"]]
        start, end = info["start_idx"], info["end_idx"]

        sliced = {}
        for key, arr in full_rec.items():
            if key.endswith(";r"):
                sliced[key] = np.asarray(arr)[start:end]
            else:
                sliced[key] = arr
        return sliced, info

    # ---------------------------------------------------------
    # Plot Unique Fish Stimuli: DF vs TF
    # ---------------------------------------------------------
    def plot_unique_fish_df_tf(self, figsize=(8, 8), marker_size=20):

        category_A = self.fish_list[0]
        category_B = self.fish_list[1]

        plt.figure(figsize=figsize)

        # Category A
        for fish in category_A:
            DF, TF, VF, MA = fish

            plt.plot(
                DF,
                TF,
                's',
                markeredgecolor='b',
                markerfacecolor='b',
                markersize=marker_size
            )

        # Category B
        for fish in category_B:
            DF, TF, VF, MA = fish

            plt.plot(
                DF,
                TF,
                '^',
                markeredgecolor='r',
                markerfacecolor='r',
                markersize=marker_size
            )

        plt.xlim(-1.1, 1.1)
        plt.ylim(-1.1, 1.1)

        plt.xlabel("Dimension 1 (DF)", fontsize=24)
        plt.ylabel("Dimension 2 (TF)", fontsize=24)

        plt.tight_layout()
        plt.show()

    # ---------------------------------------------------------
    # Plot All Pairwise Feature Spaces
    # ---------------------------------------------------------
    def plot_all_feature_pairs(
        self,
        figsize=(8, 8),
        marker_size=20,
        separate=True,
        combined=True,
        combined_figsize=(18, 11),
    ):
        features = {
            "DF": 0,
            "TF": 1,
            "VF": 2,
            "MA": 3
        }

        category_A = self.fish_list[0]
        category_B = self.fish_list[1]

        feature_pairs = [
            ("DF", "TF"),
            ("DF", "VF"),
            ("DF", "MA"),
            ("TF", "VF"),
            ("TF", "MA"),
            ("VF", "MA")
        ]

        def _draw_pair(ax, x_feature, y_feature):
            x_idx = features[x_feature]
            y_idx = features[y_feature]

            for fish in category_A:
                ax.plot(
                    fish[x_idx],
                    fish[y_idx],
                    's',
                    markeredgecolor='b',
                    markerfacecolor='b',
                    markersize=marker_size
                )

            for fish in category_B:
                ax.plot(
                    fish[x_idx],
                    fish[y_idx],
                    '^',
                    markeredgecolor='r',
                    markerfacecolor='r',
                    markersize=marker_size
                )

            ax.set_xlim(-1.1, 1.1)
            ax.set_ylim(-1.1, 1.1)

            ax.set_xlabel(f"Dimension 1 ({x_feature})", fontsize=16)
            ax.set_ylabel(f"Dimension 2 ({y_feature})", fontsize=16)

        if separate:
            for x_feature, y_feature in feature_pairs:
                fig, ax = plt.subplots(figsize=figsize)
                _draw_pair(ax, x_feature, y_feature)
                plt.tight_layout()
                plt.show()

        if combined:
            fig, axes = plt.subplots(2, 3, figsize=combined_figsize)
            axes = axes.flatten()

            for ax, (x_feature, y_feature) in zip(axes, feature_pairs):
                _draw_pair(ax, x_feature, y_feature)

            fig.tight_layout()
            plt.show()

    # ---------------------------------------------------------
    # Plot Learning Curve
    # ---------------------------------------------------------
    def plot(self, save_path="learning_trajectory.pdf", show=True):
        mean_block = np.mean(self.performance_matrix, axis=1)
        sem_block = np.std(self.performance_matrix, axis=1) / np.sqrt(self.n_repeats)
        x = np.arange(1, self.n_blocks + 1)

        plt.figure(figsize=(10, 6))
        plt.errorbar(
            x, mean_block, yerr=sem_block, fmt='-o', color='navy',
            ecolor='gray', capsize=5, linewidth=2, label='Mean Accuracy ± SEM'
        )
        plt.axhline(0.50, linestyle='--', color='red', label='Chance Level')
        plt.xlabel('Training Blocks')
        plt.ylabel('Categorization Accuracy')
        plt.title('Learning Trajectory')
        plt.legend()
        plt.grid(True, linestyle=':', alpha=0.6)
        plt.savefig(save_path, dpi=300)
        if show:
            plt.show()

    # ---------------------------------------------------------
    # Internal Plot Helper
    # ---------------------------------------------------------
    def _plot_rec(
        self,
        rec,
        populations,
        it_population,
        pfc_threshold_line,
        figsize,
        title_suffix="",
    ):
        n_plots = len(populations) + 1
        fig, axes = plt.subplots(
            nrows=n_plots, ncols=1, figsize=figsize, sharex=True
        )

        it_data = np.asarray(rec[f'{it_population};r'])
        vmin, vmax = np.min(it_data), np.max(it_data)
        axes[0].imshow(
            it_data.T, aspect='auto', vmin=vmin, vmax=vmax, cmap='viridis'
        )
        axes[0].set_title(f'{it_population} Activity (heatmap){title_suffix}')
        axes[0].set_ylabel('Neuron')

        for i, pop in enumerate(populations):
            data = np.asarray(rec[f'{pop};r'])
            n_curves = data.shape[1] if data.ndim > 1 else 1
            colors = cm.viridis(np.linspace(0, 1, n_curves))

            if n_curves == 1:
                axes[i + 1].plot(data, color=colors[0], linewidth=1.5)
            else:
                for j in range(n_curves):
                    axes[i + 1].plot(data[:, j], color=colors[j], linewidth=1.5)

            if pop == 'PFC' and pfc_threshold_line is not None:
                axes[i + 1].axhline(
                    y=pfc_threshold_line, color='red', linestyle='--', linewidth=2
                )

            axes[i + 1].set_title(f'{pop} Activity')
            axes[i + 1].set_ylabel('Activity (r)')
            axes[i + 1].grid(True)

        axes[-1].set_xlabel('Time (ms)')
        plt.tight_layout()
        plt.show()
        return fig, axes

    # ---------------------------------------------------------
    # Plot Full Chunk
    # ---------------------------------------------------------
    def plot_population_recordings(
        self,
        chunk_idx,
        populations=None,
        it_population="IT",
        pfc_threshold_line=1.0,
        figsize=(10, 14),
    ):
        if populations is None:
            populations = ['StrD1', 'StrD2', 'STN', 'GPe', 'GPi',
                           'Thal', 'PFC', 'PPN', 'SNc']

        recordings = self.mons.get_recordings()
        rec = recordings[chunk_idx]
        return self._plot_rec(
            rec, populations, it_population, pfc_threshold_line, figsize
        )

    # ---------------------------------------------------------
    # Plot Single Trial
    # ---------------------------------------------------------
    def plot_trial_recording(
        self,
        block_idx,
        trial_idx,
        repeat_idx=0,
        populations=None,
        it_population="IT",
        pfc_threshold_line=1.0,
        figsize=(10, 14),
    ):
        if populations is None:
            populations = ['StrD1', 'StrD2', 'STN', 'GPe', 'GPi',
                           'Thal', 'PFC', 'PPN', 'SNc']

        rec, info = self.get_trial_recording(block_idx, trial_idx, repeat_idx)
        status = "correct" if info["correct"] else "incorrect"
        title_suffix = (
            f" — Block {block_idx+1}, Trial {trial_idx+1} "
            f"({status}, true={info['category_true']}, "
            f"selected={info['category_selected']})"
        )
        return self._plot_rec(
            rec, populations, it_population, pfc_threshold_line, figsize,
            title_suffix=title_suffix,
        )

    # ---------------------------------------------------------
    # Get Misclassified Trials
    # ---------------------------------------------------------
    def get_misclassified_trials(self):
        """لیستی از تمام تریال‌ها و ماهی‌هایی که به اشتباه دسته‌بندی شده‌اند را برمی‌گرداند."""
        misclassified_list = []

        for block_idx in range(self.n_blocks):
            for repeat_idx in range(self.n_repeats):
                trials = self.trial_log[block_idx][repeat_idx]
                for trial_idx, trial_info in enumerate(trials):
                    if trial_info["correct"] == 0:
                        error_record = {
                            "block_idx": block_idx,
                            "repeat_idx": repeat_idx,
                            "trial_idx": trial_idx,
                            "fish": trial_info["fish"],
                            "category_true": trial_info["category_true"],
                            "category_selected": trial_info["category_selected"],
                            "decided_at_feature": trial_info["decided_at_feature"],
                        }
                        misclassified_list.append(error_record)

        return misclassified_list

    # ---------------------------------------------------------
    # Analyze Misclassifications
    # ---------------------------------------------------------
    def analyze_misclassifications(self):
        """خطاهای مدل را بر اساس ویژگی‌های ماهی و نوع خطا گروه‌بندی و تحلیل می‌کند."""
        wrong_trials = self.get_misclassified_trials()

        error_patterns = defaultdict(int)
        confusion_matrix_errors = defaultdict(int)

        for err in wrong_trials:
            fish = err["fish"]
            true_cat = err["category_true"]
            pred_cat = err["category_selected"]

            fish_signature = tuple(sorted(fish.items()))

            error_patterns[fish_signature] += 1
            confusion_matrix_errors[(true_cat, str(pred_cat))] += 1

        print(
            f"\n--- آمار کلی: {len(wrong_trials)} خطای ثبت‌شده در کل آزمایش ---"
        )

        print("\n[۱] بررسی جابجایی دسته‌ها (True vs Predicted):")
        for (true_c, pred_c), count in confusion_matrix_errors.items():
            print(f"  ماهیِ واقعی '{true_c}' که اشتباهی تشخیص داده شد '{pred_c}': {count} بار")

        print("\n[۲] پرتکرارترین ترکیب ویژگی‌های ماهی که مدل را به اشتباه انداخته:")
        sorted_patterns = sorted(
            error_patterns.items(), key=lambda x: x[1], reverse=True
        )

        for fish_sig, count in sorted_patterns[:5]:
            fish_dict = dict(fish_sig)
            print(f"  ترکیب ویژگی‌ها: {fish_dict} --> تعداد تکرار خطا: {count} بار")

        return error_patterns, confusion_matrix_errors
```

```python
def lock_params():
    """
    ابتدا پارامترها را اعمال کرده و سپس وضعیت فعلی آن‌ها
    (K_burstها، آستانه‌ها و بیس‌لاین SNc) را در یک دیکشنری قفل و ذخیره می‌کند.
    """
    # ۱. اعمال پارامترها
    apply_network_parameters()

    # ۲. ثبت و قفل کردن مقادیر فعلی پروژکشن‌ها و متغیرهای گلوبال
    locked_state = {
        "projections": {
            "ITStrD1": {
                "K_burst": ITStrD1.K_burst,
                "threshold_pre": ITStrD1.threshold_pre
            },
            "StrD1GPi": {
                "K_burst": StrD1GPi.K_burst,
                "K_dip": StrD1GPi.K_dip,
                "threshold_pre": StrD1GPi.threshold_pre,
                "threshold_post": StrD1GPi.threshold_post
            },
            "ITStrD2": {
                "K_burst": ITStrD2.K_burst,
                "K_dip": ITStrD2.K_dip,
                "threshold_pre": ITStrD2.threshold_pre
            },
            "StrD2GPe": {
                "K_burst": StrD2GPe.K_burst,
                "K_dip": StrD2GPe.K_dip
            },
            "GPiGPi": {
                "w": GPiGPi.w
            }
        },
        "globals": {
            "SNc_baseline": SNc_baseline
        }
    }

    print("🔒 Parameters locked successfully! اسنپ‌شات پارامترها با موفقیت ذخیره شد.")
    return locked_state

# فراخوانی و فریز کردن پارامترها
current_locked_params = lock_params()
```

```python

#--------------control task---------
#for every block a specific presentation order will be fixed
from collections import deque
import numpy as np
import random
import itertools
import matplotlib.pyplot as plt
import matplotlib.cm as cm

from ANNarchy import simulate, reset
from CompNeuroPy import CompNeuroMonitors, get_full_model

try:
    from ANNarchy import dt as _ann_dt
except ImportError:
    _ann_dt = None


class CategorizationExperiment:

    # SETUP MONITORS
    # =========================================================
    @staticmethod
    def setup_monitors(verbose=True):

        full_model = get_full_model()
        populations = full_model["populations"]

        monitor_config = {
            pop_name: ["r"]
            for pop_name in populations
        }

        mons = CompNeuroMonitors(monitor_config)

        if verbose:
            print(
                f"Monitors initialized for "
                f"{len(populations)} populations."
            )
            print(
                f"Monitoring configuration: "
                f"{list(monitor_config.keys())}"
            )

        return mons, populations


    # =========================================================
    # INITIALIZATION
    # =========================================================
    def __init__(
        self,
        mons,
        IT,
        PFC,
        PPN,
        SNc,
        StrD1GPi,
        StrD2GPe,
        fish_list,
        feature_name_list,
        get_fish,
        get_true_category,
        get_feature_order,
        select_input,
        n_blocks=12,
        n_repeats=2,
        nr_fish_presentations=150,
        trial_duration=210,
        settle_duration=30,
        dopamine_input_duration=245,
        baseline_it=0.0,
        response_threshold=1.3,
        margin_threshold=0.2,
        seed=44,
        dt=None,
        weight_clip_min=0.05,
        weight_clip_max=3.0,
        dopamine_correct=1.0,
        dopamine_incorrect=0.05,
    ):

        self.mons = mons

        self.IT = IT
        self.PFC = PFC
        self.PPN = PPN
        self.SNc = SNc
        self.StrD1GPi = StrD1GPi
        self.StrD2GPe = StrD2GPe

        self.fish_list = fish_list
        self.feature_name_list = feature_name_list

        self.get_fish = get_fish
        self.get_true_category = get_true_category
        self.get_feature_order = get_feature_order
        self.select_input = select_input

        self.n_blocks = n_blocks
        self.n_repeats = n_repeats
        self.nr_fish_presentations = nr_fish_presentations

        self.trial_duration = trial_duration
        self.settle_duration = settle_duration
        self.dopamine_input_duration = dopamine_input_duration

        self.baseline_it = baseline_it

        self.response_threshold = response_threshold
        self.margin_threshold = margin_threshold

        self.weight_clip_min = weight_clip_min
        self.weight_clip_max = weight_clip_max

        self.dopamine_correct = dopamine_correct
        self.dopamine_incorrect = dopamine_incorrect

        self.seed = seed

        np.random.seed(seed)
        random.seed(seed)

        # -----------------------------------------------------
        # ANNarchy dt
        # -----------------------------------------------------
        if dt is not None:
            self.dt = dt

        elif _ann_dt is not None:
            try:
                self.dt = _ann_dt()
            except Exception:
                self.dt = 1.0

        else:
            self.dt = 1.0

        # -----------------------------------------------------
        # Performance matrix
        # -----------------------------------------------------
        self.performance_matrix = np.zeros(
            (n_blocks, n_repeats)
        )

        # -----------------------------------------------------
        # Trial log
        # -----------------------------------------------------
        self.trial_log = {
            b: {
                r: []
                for r in range(n_repeats)
            }
            for b in range(n_blocks)
        }

        # -----------------------------------------------------
        # Fixed feature order for each block
        # -----------------------------------------------------
        self.block_feature_orders = {}

        # -----------------------------------------------------
        # Recording chunk counter
        # -----------------------------------------------------
        self._chunk_counter = 0


    # =========================================================
    # FEATURE ORDERS
    # =========================================================
    def create_feature_orders(self):

        """
        Create 12 unique feature permutations.

        Four features:
            DF
            VF
            TF
            MA

        Each feature appears exactly 3 times
        in the first position.
        """

        feature_orders = [

            # -------------------------------------------------
            # DF starts
            # -------------------------------------------------
            ['DF', 'TF', 'VF', 'MA'],
            ['DF', 'MA', 'TF', 'VF'],
            ['DF', 'VF', 'MA', 'TF'],

            # -------------------------------------------------
            # VF starts
            # -------------------------------------------------
            ['VF', 'DF', 'TF', 'MA'],
            ['VF', 'MA', 'DF', 'TF'],
            ['VF', 'TF', 'MA', 'DF'],

            # -------------------------------------------------
            # TF starts
            # -------------------------------------------------
            ['TF', 'DF', 'MA', 'VF'],
            ['TF', 'VF', 'DF', 'MA'],
            ['TF', 'MA', 'VF', 'DF'],

            # -------------------------------------------------
            # MA starts
            # -------------------------------------------------
            ['MA', 'DF', 'VF', 'TF'],
            ['MA', 'TF', 'DF', 'VF'],
            ['MA', 'VF', 'TF', 'DF'],
        ]

        # -----------------------------------------------------
        # Validate number of blocks
        # -----------------------------------------------------
        if self.n_blocks > len(feature_orders):
            raise ValueError(
                f"n_blocks={self.n_blocks}, "
                f"but only {len(feature_orders)} "
                f"feature orders are available."
            )

        # -----------------------------------------------------
        # Validate uniqueness
        # -----------------------------------------------------
        tuples = [
            tuple(order)
            for order in feature_orders
        ]

        if len(tuples) != len(set(tuples)):
            raise ValueError(
                "Duplicate feature orders detected."
            )

        # -----------------------------------------------------
        # Validate that each feature starts 3 times
        # -----------------------------------------------------
        first_features = [
            order[0]
            for order in feature_orders
        ]

        for feature in self.feature_name_list:

            count = first_features.count(feature)

            if count != 3:
                raise ValueError(
                    f"{feature} appears {count} times "
                    f"in first position instead of 3."
                )

        return feature_orders


    # =========================================================
    # SINGLE TRIAL
    # =========================================================
    def run_trial(self, feature_order):

        simulate(120)

        fish = self.get_fish(
            self.fish_list,
            self.feature_name_list
        )

        category_true = self.get_true_category(
            fish
        )

        category_selected = None
        trial_correct = 0

        decided_at_feature = None

        resp_val_final = None
        margin_final = None

        self.mons.start()

        attempts_used = 0

        # -----------------------------------------------------
        # Use fixed block-specific feature order
        # -----------------------------------------------------
        for idx, feature_name in enumerate(
            feature_order,
            start=1
        ):

            attempts_used = idx

            self.select_input(
                feature_name,
                fish,
                self.IT,
                self.feature_name_list,
                self.baseline_it
            )

            simulate(
                self.trial_duration
                + self.settle_duration
            )

            # -------------------------------------------------
            # PFC response
            # -------------------------------------------------
            pfc_act = np.array(
                self.PFC.r
            ).flatten()

            resp_val = np.max(
                pfc_act
            )

            resp_idx = np.argmax(
                pfc_act
            )

            sorted_pfc = np.sort(
                pfc_act
            )

            margin = (
                sorted_pfc[-1]
                - sorted_pfc[-2]
            )

            # -------------------------------------------------
            # Dynamic threshold
            # -------------------------------------------------
            dynamic_thresh = (
                0.8
                if idx == len(feature_order)
                else self.response_threshold
            )

            # -------------------------------------------------
            # Decision
            # -------------------------------------------------
            if (
                resp_val >= dynamic_thresh
                and
                margin >= self.margin_threshold
            ):

                category_selected = [
                    "A",
                    "B"
                ][resp_idx]

                trial_correct = int(
                    category_selected
                    == category_true
                )

                decided_at_feature = (
                    feature_name
                )

                resp_val_final = (
                    resp_val
                )

                margin_final = (
                    margin
                )

                # -------------------------------------------------
                # Dopamine feedback
                # -------------------------------------------------
                if trial_correct:

                    self.PPN.B = 1.0
                    self.SNc.alpha = (
                        self.dopamine_correct
                    )

                else:

                    self.PPN.B = 0.0
                    self.SNc.alpha = (
                        self.dopamine_incorrect
                    )

                simulate(
                    self.dopamine_input_duration
                )

                # -------------------------------------------------
                # Weight clipping
                # -------------------------------------------------
                self.StrD1GPi.w = np.clip(
                    self.StrD1GPi.w,
                    self.weight_clip_min,
                    self.weight_clip_max
                )

                self.StrD2GPe.w = np.clip(
                    self.StrD2GPe.w,
                    self.weight_clip_min,
                    self.weight_clip_max
                )

                break

        self.mons.pause()

        # -----------------------------------------------------
        # Reset inputs
        # -----------------------------------------------------
        self.IT.B = 0
        self.PPN.B = 0
        self.SNc.alpha = 0

        simulate(120)

        # -----------------------------------------------------
        # Recording duration
        # -----------------------------------------------------
        recorded_ms = (
            attempts_used
            * (
                self.trial_duration
                + self.settle_duration
            )
        )

        if decided_at_feature is not None:
            recorded_ms += (
                self.dopamine_input_duration
            )

        # -----------------------------------------------------
        # Trial information
        # -----------------------------------------------------
        trial_info = {

            "fish": fish,

            "category_true":
                category_true,

            "category_selected":
                category_selected,

            "correct":
                trial_correct,

            "feature_order":
                list(feature_order),

            "decided_at_feature":
                decided_at_feature,

            "response_value":
                resp_val_final,

            "margin":
                margin_final,

            "recorded_ms":
                recorded_ms,

            "n_samples":
                int(
                    round(
                        recorded_ms
                        / self.dt
                    )
                ),
        }

        return trial_info


    # =========================================================
    # REPEAT
    # =========================================================
    def run_repeat(
        self,
        block_idx,
        repeat_idx,
        feature_order
    ):

        self.mons.reset()

        chunk_idx = (
            self._chunk_counter
        )

        self._chunk_counter += 1

        offset = 0

        block_accuracies = deque()

        # -----------------------------------------------------
        # Trials
        # -----------------------------------------------------
        for trial_idx in range(
            self.nr_fish_presentations
        ):

            trial_info = self.run_trial(
                feature_order
            )

            trial_info["trial_idx"] = (
                trial_idx
            )

            trial_info["chunk_idx"] = (
                chunk_idx
            )

            trial_info["start_idx"] = (
                offset
            )

            trial_info["end_idx"] = (
                offset
                + trial_info["n_samples"]
            )

            offset = (
                trial_info["end_idx"]
            )

            self.trial_log[
                block_idx
            ][
                repeat_idx
            ].append(
                trial_info
            )

            block_accuracies.append(
                trial_info["correct"]
            )

        return float(
            np.mean(
                block_accuracies
            )
        )


    # =========================================================
    # FULL RUN
    # =========================================================
    def run(self, verbose=True):

        np.random.seed(
            self.seed
        )

        random.seed(
            self.seed
        )

        # -----------------------------------------------------
        # Create fixed feature orders
        # -----------------------------------------------------
        feature_orders = (
            self.create_feature_orders()
        )

        # -----------------------------------------------------
        # Run blocks
        # -----------------------------------------------------
        for block_idx in range(
            self.n_blocks
        ):

            if verbose:

                print(
                    f"\n===== Block "
                    f"{block_idx + 1}"
                    f"/{self.n_blocks} ====="
                )

            # -------------------------------------------------
            # Reset neural model
            # -------------------------------------------------
            reset()

            simulate(50)

            # -------------------------------------------------
            # Fixed order for this block
            # -------------------------------------------------
            feature_order = (
                feature_orders[
                    block_idx
                ]
            )

            # -------------------------------------------------
            # Store order
            # -------------------------------------------------
            self.block_feature_orders[
                block_idx
            ] = feature_order.copy()

            if verbose:

                print(
                    "  Feature order: "
                    + " → ".join(
                        feature_order
                    )
                )

            # -------------------------------------------------
            # Repeats
            # -------------------------------------------------
            for repeat in range(
                self.n_repeats
            ):

                acc = self.run_repeat(
                    block_idx,
                    repeat,
                    feature_order
                )

                self.performance_matrix[
                    block_idx,
                    repeat
                ] = acc

                if verbose:

                    print(
                        f"  Repeat "
                        f"{repeat + 1} "
                        f"accuracy: "
                        f"{acc:.3f}"
                    )

        return (
            self.performance_matrix
        )


    # =========================================================
    # GET TRIAL INFO
    # =========================================================
    def get_trial_info(
        self,
        block_idx,
        trial_idx,
        repeat_idx=0
    ):

        try:

            return (
                self.trial_log[
                    block_idx
                ][
                    repeat_idx
                ][
                    trial_idx
                ]
            )

        except (
            KeyError,
            IndexError
        ):

            raise ValueError(
                "Invalid block/repeat/trial index."
            )


    # =========================================================
    # GET BLOCK INFO
    # =========================================================
    def get_block_info(
        self,
        block_idx,
        repeat_idx=0
    ):

        try:

            return (
                self.trial_log[
                    block_idx
                ][
                    repeat_idx
                ]
            )

        except KeyError:

            raise ValueError(
                "Invalid block/repeat index."
            )


    # =========================================================
    # GET BLOCK FEATURE ORDER
    # =========================================================
    def get_block_feature_order(
        self,
        block_idx
    ):

        try:

            return self.block_feature_orders[
                block_idx
            ]

        except KeyError:

            raise ValueError(
                "Invalid block index."
            )


    # =========================================================
    # GET TRIAL RECORDING
    # =========================================================
    def get_trial_recording(
        self,
        block_idx,
        trial_idx,
        repeat_idx=0
    ):

        info = self.get_trial_info(
            block_idx,
            trial_idx,
            repeat_idx
        )

        recordings = (
            self.mons.get_recordings()
        )

        full_rec = recordings[
            info["chunk_idx"]
        ]

        start = info[
            "start_idx"
        ]

        end = info[
            "end_idx"
        ]

        sliced = {}

        for key, arr in full_rec.items():

            if key.endswith(";r"):

                sliced[key] = np.asarray(
                    arr
                )[start:end]

            else:

                sliced[key] = arr

        return (
            sliced,
            info
        )


    # =========================================================
    # PLOT LEARNING CURVE
    # =========================================================
    def plot(
        self,
        save_path="learning_trajectory.pdf",
        show=True
    ):

        mean_block = np.mean(
            self.performance_matrix,
            axis=1
        )

        sem_block = (
            np.std(
                self.performance_matrix,
                axis=1
            )
            / np.sqrt(
                self.n_repeats
            )
        )

        x = np.arange(
            1,
            self.n_blocks + 1
        )

        plt.figure(
            figsize=(10, 6)
        )

        plt.errorbar(
            x,
            mean_block,
            yerr=sem_block,
            fmt='-o',
            color='navy',
            ecolor='gray',
            capsize=5,
            linewidth=2,
            label='Mean Accuracy ± SEM'
        )

        plt.axhline(
            0.50,
            linestyle='--',
            color='red',
            label='Chance Level'
        )

        plt.xlabel(
            'Training Blocks'
        )

        plt.ylabel(
            'Categorization Accuracy'
        )

        plt.title(
            'Learning Trajectory'
        )

        plt.legend()

        plt.grid(
            True,
            linestyle=':',
            alpha=0.6
        )

        plt.savefig(
            save_path,
            dpi=300
        )

        if show:
            plt.show()


    # =========================================================
    # INTERNAL PLOT HELPER
    # =========================================================
    def _plot_rec(
        self,
        rec,
        populations,
        it_population,
        pfc_threshold_line,
        figsize,
        title_suffix=""
    ):

        n_plots = (
            len(populations)
            + 1
        )

        fig, axes = plt.subplots(
            nrows=n_plots,
            ncols=1,
            figsize=figsize,
            sharex=True
        )

        # -----------------------------------------------------
        # IT heatmap
        # -----------------------------------------------------
        it_data = np.asarray(
            rec[
                f'{it_population};r'
            ]
        )

        vmin = np.min(
            it_data
        )

        vmax = np.max(
            it_data
        )

        axes[0].imshow(
            it_data.T,
            aspect='auto',
            vmin=vmin,
            vmax=vmax,
            cmap='viridis'
        )

        axes[0].set_title(
            f'{it_population} '
            f'Activity (heatmap)'
            f'{title_suffix}'
        )

        axes[0].set_ylabel(
            'Neuron'
        )

        # -----------------------------------------------------
        # Population recordings
        # -----------------------------------------------------
        for i, pop in enumerate(
            populations
        ):

            data = np.asarray(
                rec[
                    f'{pop};r'
                ]
            )

            n_curves = (
                data.shape[1]
                if data.ndim > 1
                else 1
            )

            colors = cm.viridis(
                np.linspace(
                    0,
                    1,
                    n_curves
                )
            )

            if n_curves == 1:

                axes[i + 1].plot(
                    data,
                    color=colors[0],
                    linewidth=1.5
                )

            else:

                for j in range(
                    n_curves
                ):

                    axes[i + 1].plot(
                        data[:, j],
                        color=colors[j],
                        linewidth=1.5
                    )

            if (
                pop == 'PFC'
                and
                pfc_threshold_line
                is not None
            ):

                axes[i + 1].axhline(
                    y=pfc_threshold_line,
                    color='red',
                    linestyle='--',
                    linewidth=2
                )

            axes[i + 1].set_title(
                f'{pop} Activity'
            )

            axes[i + 1].set_ylabel(
                'Activity (r)'
            )

            axes[i + 1].grid(
                True
            )

        axes[-1].set_xlabel(
            'Time (ms)'
        )

        plt.tight_layout()

        plt.show()

        return (
            fig,
            axes
        )


    # =========================================================
    # PLOT FULL CHUNK
    # =========================================================
    def plot_population_recordings(
        self,
        chunk_idx,
        populations=None,
        it_population="IT",
        pfc_threshold_line=1.0,
        figsize=(10, 14)
    ):

        if populations is None:

            populations = [
                'StrD1',
                'StrD2',
                'STN',
                'GPe',
                'GPi',
                'Thal',
                'PFC',
                'PPN',
                'SNc'
            ]

        recordings = (
            self.mons.get_recordings()
        )

        rec = recordings[
            chunk_idx
        ]

        return self._plot_rec(
            rec,
            populations,
            it_population,
            pfc_threshold_line,
            figsize
        )


    # =========================================================
    # PLOT SINGLE TRIAL
    # =========================================================
    def plot_trial_recording(
        self,
        block_idx,
        trial_idx,
        repeat_idx=0,
        populations=None,
        it_population="IT",
        pfc_threshold_line=1.0,
        figsize=(10, 14)
    ):

        if populations is None:

            populations = [
                'StrD1',
                'StrD2',
                'STN',
                'GPe',
                'GPi',
                'Thal',
                'PFC',
                'PPN',
                'SNc'
            ]

        rec, info = (
            self.get_trial_recording(
                block_idx,
                trial_idx,
                repeat_idx
            )
        )

        status = (
            "correct"
            if info["correct"]
            else "incorrect"
        )

        title_suffix = (
            f" — Block "
            f"{block_idx + 1}, "
            f"Trial "
            f"{trial_idx + 1} "
            f"({status}, "
            f"true="
            f"{info['category_true']}, "
            f"selected="
            f"{info['category_selected']}, "
            f"features="
            f"{info['feature_order']})"
        )

        return self._plot_rec(
            rec,
            populations,
            it_population,
            pfc_threshold_line,
            figsize,
            title_suffix=title_suffix
        )
```
